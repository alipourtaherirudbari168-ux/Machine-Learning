```python
import numpy as np
import pandas as pd
# صدا زدن کتابخانه ها
```

```python
movies_df = pd.DataFrame(pd.read_csv('movies.csv'))
ratings_df = pd.DataFrame(pd.read_csv('ratings.csv'))
# خواندن فایل ها 
```

```python
movies_df.head()
# نشون دادن فایل فیلم ها
```

```python
movies_df['year'] = movies_df['title'].str.extract(r'(\(\d+\))', expand = False)
movies_df['year'] = movies_df['year'].str.extract(r'(\d+)', expand = False)
# با این کد یک ویژگی جدید به اسم سال درست کردم که تاریخ رو نگه داره 

movies_df['title'] = movies_df['title'].replace(r'(\(\d+\))', '', regex = True).str.strip()
# اینجا هم اون زمان رو از اسم فیلم ها پاک کردم و با استریپ فاصله های اضافی رو حذف کردم

movies_df = movies_df[movies_df['genres'] != '(no genres listed)'].reset_index(drop = True)
# اون هایی که ندارن ژانر‌ رو حذف کردم 

movies_df['genres'] = movies_df['genres'].str.split('|')
# سبک هارو به لیست تبدیل کردم که بعد بتونم مولتی لیبل باینریزر کنم
```

```python
movies_df.head()
# فایل فیلم هارو به صورت دیتا فریم نهایی نشون دادم
```

```python
from sklearn.preprocessing import MultiLabelBinarizer
# کتابخانه مورد نظر رو صدا زدم
mlb = MultiLabelBinarizer()
# این خط مثل کلاس ها میمونه که یک نمونه ساختم
genres_encoded = mlb.fit_transform(movies_df['genres'])
# اینجا یک لیست ساختم که بگه فیلم ها هرکدوم در هر سبک هستند یا نه
MoviesGenres = pd.DataFrame(genres_encoded, columns = mlb.classes_, index = movies_df.index)
#اینجا گفتم از چیزی که بالا ساختی با ستون های اسم سبک ها دیتا فریم بساز 
```

```python
MoviesGenres.head()
# اینجا هم نشون دادم که چجوری شد اون چیزی که برای سبک ها ساختم
```

```python
MoviesWithGenres = pd.concat([movies_df, MoviesGenres], axis = 1)
# اینجا هم اون چیزی که از سبک ها ساختم رو با اون فیلم ها وصل کردم بهم به صورت ستونی
```

```python
MoviesWithGenres.head()
# اینجا هم باز نشون دادم چیزی که بالا هست
```

```python
ratings_df = ratings_df.drop('timestamp', axis = 1)
# از امتیاز ها زمان ساخت رو استفاده نکردم
```

```python
ratings_df.head()
# نشون دادم اون دیتا فریم امتیاز هارو
```

```python
userInput = [
              {'title': 'Breakfast Club, The', 'rating': 5},
              {'title': 'Toy Story', 'rating': 3.5},
              {'title': 'Jumanji', 'rating': 2},
              {'title': 'Pulp Fiction', 'rating': 5},
              {'title': 'Akira', 'rating': 4.5}
           ]

inputMovies = pd.DataFrame(userInput)

# اینجا مثلا امیتاز های یک شخص هست 
```

```python
inputMovies.head()
# اینجا نشون دادم چیزی که بالا هست
```

```python
inputId = MoviesWithGenres[MoviesWithGenres['title'].isin(inputMovies['title'].tolist())].reset_index(drop = True)
# اینجا از اون فیلم هایی که با سبکشون بود اونایی که کاربر بهش نمره داده رو جدا کردم

inputMovies = pd.merge(inputId, inputMovies)
inputMovies = inputMovies.drop(['genres', 'year', 'title', 'movieId'], axis = 1).reset_index(drop = True)
# اینجا هم بعضی از ویژگی هارو استفاده نکردم و قبلش باز وصل کردم دیتار فریم همراه با شماره فیلم و دیتارفریم فیلم هارو که دیتا فریمی از سبک ها و نمره داشته باشم
```

```python
inputMovies.head()
```

```python
userGenres = inputMovies.drop('rating', axis = 1)
# اینجا نمره هارو استفاده نکردم و یک دیتا فریم فقط با نوع فیلم هست 
```

```python
userGenres.head()
```

```python
userProfile = userGenres.transpose().dot(inputMovies['rating'])
# ضرب ماتریسی واسه اینکه بفهمیم کدوم نوع فیلم هارو کاربر نمره چجوری داده
```

```python
userProfile
```

```python
genresTable = MoviesWithGenres.set_index(MoviesWithGenres['movieId']).drop(['genres', 'year', 'title', 'movieId'], axis = 1)
# شماره فیلم رو در جای اون صفر یک دو و غیره گذاشتیم 
```

```python
genresTable.head()
```

```python
rec = ((genresTable * userProfile).sum(axis = 1)/(userProfile.sum()))
# اینجا هم ابنکه هر فیلم رو چقدر میپسنده کاربر مشخص کردم 
```

```python
rec = rec.sort_values(ascending = False)
# از زیاد به کم مرتب کردم
```

```python
rec.head(10)
```

```python
finally_rec = movies_df[movies_df['movieId'].isin(rec.head(10).index.tolist())].drop(['genres'], axis = 1).reset_index(drop = True)
#اینجا ده تا از فیلم هارو که در نهایت بیشترین شباهت رو داشتن رو از فیلم ها برداشتیم 
```
