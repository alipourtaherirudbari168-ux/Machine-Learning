```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(-9, 9, 0.1)

y = x**3 + x**2 + x + 1
y_noise = 20*np.random.normal(size = x.size)
Y = y + y_noise

plt.plot(x, y, 'r')

plt.scatter(x, Y, color = 'royalblue', edgecolor = 'k')
```

```python
x = np.arange(-9, 9, 0.1)

y = np.pow(x, 2)
y_noise = 3*np.random.normal(size = x.size)

Y = y + y_noise

plt.plot(x, y, 'r')

plt.scatter(x, Y, color = 'royalblue', edgecolor = 'k')
```

```python
x = np.arange(-9, 9, 0.1)
y = np.exp(x)

y_noise = 40*np.random.normal(size = x.size)
Y = y + y_noise

plt.plot(x, y, 'k--')
```

```python
x = np.arange(-10, 10, 0.1)

y = 1 - ( 4 / ( 1 + np.power(3, x - 2) ) )

plt.plot(x, y, 'k--')

# Sigmoid: 1 / (1 + np.exp( B1( X - B2) ) )
```

```python
x = np.arange(0.1, 10, 0.1)

y = np.log(x)

plt.plot(x, y, 'k--')
```
